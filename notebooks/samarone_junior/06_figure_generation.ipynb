# 06 — Publication Figure Generation

This notebook uses the `visualization.figures` module to generate all 8 publication-quality PDF figures for the LaTeX report.  
Each figure function reads experiment results from `metrics.jsonl` (except the circuit and pipeline schematics) and produces a PDF.

## 1. Setup

We import all figure functions and load the results DataFrame.

````python
import pandas as pd
from pathlib import Path
from IPython.display import display, Image

from qml.samarone_junior.visualization.figures import (
    fig_qae_circuit,
    fig_preprocessing_pipeline,
    fig_roc_curves,
    fig_pr_curves,
    fig_threshold_histograms,
    fig_f1_boxplots,
    fig_param_vs_f1,
    fig_event_heatmap,
)

# Load experiment results
results_path = "../../data/samarone_junior/results/metrics.jsonl"
df = pd.read_json(results_path, lines=True)
print(f"Loaded {len(df)} result rows from {results_path}")

# Output directory
out = Path("../../data/samarone_junior/figures")
out.mkdir(parents=True, exist_ok=True)
````

## 2. Schematic Figures

These figures do not depend on experiment results — they illustrate the QAE circuit and preprocessing pipeline.

````python
# Figure 1: QAE Circuit Diagram
fig_qae_circuit(out / "fig_qae_circuit.pdf")
print("✓ fig_qae_circuit.pdf")
````

````python
# Figure 2: Preprocessing Pipeline Schematic
fig_preprocessing_pipeline(out / "fig_preprocessing_pipeline.pdf")
print("✓ fig_preprocessing_pipeline.pdf")
````

## 3. Metric Comparison Figures

These figures aggregate the cross-validated results to compare model performance.

````python
# Figure 3: AUC-ROC Bar Chart
fig_roc_curves(df, out / "fig_roc_curves.pdf")
print("✓ fig_roc_curves.pdf")
````

````python
# Figure 4: AUC-PR Bar Chart
fig_pr_curves(df, out / "fig_pr_curves.pdf")
print("✓ fig_pr_curves.pdf")
````

````python
# Figure 5: Threshold Value Histograms
fig_threshold_histograms(df, out / "fig_threshold_histograms.pdf")
print("✓ fig_threshold_histograms.pdf")
````

````python
# Figure 6: F1 Score Boxplots
fig_f1_boxplots(df, out / "fig_f1_boxplots.pdf")
print("✓ fig_f1_boxplots.pdf")
````

````python
# Figure 7: Parameter Count vs F1 (R018)
fig_param_vs_f1(df, out / "fig_param_vs_f1.pdf")
print("✓ fig_param_vs_f1.pdf")
````

````python
# Figure 8: Event-Type Heatmap
fig_event_heatmap(df, out / "fig_event_heatmap.pdf")
print("✓ fig_event_heatmap.pdf")
````

## 4. Verify Generated Figures

List all generated PDFs and confirm file sizes are non-trivial.

````python
import os

for p in sorted(out.glob("*.pdf")):
    size_kb = os.path.getsize(p) / 1024
    status = "✓" if size_kb > 1 else "✗"
    print(f"  {status} {p.name:40s}  {size_kb:.1f} KB")
````
